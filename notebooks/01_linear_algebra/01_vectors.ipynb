## 1. What is a vector?

A **vector** is an ordered collection of components. In elementary numerical work, those components are often real numbers.

For example:

$$
\mathbf{v} = \begin{bmatrix}2\\3\end{bmatrix} = (2,3)
$$

The vector has two components, so its dimension is 2. In two-dimensional geometry, it can be pictured as an arrow from the origin to the point $(2,3)$.

### Scalar, vector, and matrix

| Object | Example | Interpretation |
|---|---|---|
| Scalar | `3.5` | One numerical value |
| Vector | `[2, 3]` | An ordered collection of values |
| Matrix | `[[1, 2], [3, 4]]` | Values arranged in rows and columns |

The order matters. In general, $[2,3]$ and $[3,2]$ represent different vectors.

### Components and indexing

For $\mathbf{v}=[2,3,4]$, the components are 2, 3, and 4. In mathematical notation, components are often written $v_1,v_2,v_3$. In Python, sequence indexing starts at 0, so the same values are accessed at indices 0, 1, and 2.

### Machine-learning connection

A vector can represent one observation's numerical features, a collection of model parameters, or an embedding produced by a model. The meaning of each position must remain consistent across observations.

**Check your understanding:** What is the dimension of $[10,20,30]$? What does each component mean in a problem you care about?

```python
v = [2.0, 3.0]
print("Vector:", v)
print("Number of components:", len(v))
```

## 2. Dimension, shape, and magnitude are different

These terms are related, but they are not interchangeable.

- **Dimension** (in the vector-space sense, for a single vector) tells us how many components the vector has.
- **Shape** describes the structure of an array in a programming library such as NumPy.
- **Magnitude** or the Euclidean norm measures the vector's length.

For $\mathbf{v}=[2,3]$, its Euclidean magnitude is

$$
\|\mathbf{v}\|_2 = \sqrt{2^2+3^2}=\sqrt{13}\approx 3.606.
$$

A two-component vector has dimension 2 regardless of whether its magnitude is small or large.

### Shape conventions in NumPy

| Representation | Shape | Meaning |
|---|---|---|
| `np.array([2, 3])` | `(2,)` | One-dimensional array with two elements |
| `np.array([[2], [3]])` | `(2, 1)` | Two rows and one column |
| `np.array([[2, 3]])` | `(1, 2)` | One row and two columns |

These arrays contain related values, but their shapes affect broadcasting and matrix operations. Shape bugs are a common source of ML implementation errors.

**Your prediction:** before running the next cell, write down what you expect `len(v)` and a future NumPy version of `v`'s shape to be.

```python
from math import sqrt

magnitude = sqrt(sum(component ** 2 for component in v))
print("Vector:", v)
print("Number of components:", len(v))
print("Magnitude:", magnitude)
```

## 3. Vector addition and subtraction

Let

$$
\mathbf{a}=[2,3],\qquad \mathbf{b}=[1,4].
$$

Addition and subtraction operate **component by component**:

$$
\mathbf{a}+\mathbf{b}=[2+1,\,3+4]=[3,7]
$$

$$
\mathbf{a}-\mathbf{b}=[2-1,\,3-4]=[1,-1].
$$

Both vectors must have the same number of components for these ordinary operations. Adding a two-component vector to a three-component vector is not defined as standard component-wise vector addition.

### A Python detail worth remembering

Python lists are general-purpose containers. The expression `[2, 3] + [1, 4]` concatenates lists rather than adding corresponding components. You will implement the mathematical behavior explicitly.

### Your turn

First compute the two results by hand. Then implement `add_vectors(a, b)` and `subtract_vectors(a, b)` using a loop or a list comprehension. Include a dimension check before combining the components.

```python
a = [2, 3]
b = [1, 4]

def add_vectors(a, b):
    """Return the component-wise sum of thwo vectors."""
    if len(a) != len(b):
        raise ValueError("Vectos must have the same length.")

    return [x + y for x, y in zip(a, b)]

def subtract_vectors(a, b):
    """Return the component-wise sum of thwo vectors."""
    if len(a) != len(b):
        raise ValueError("Vectos must have the same length.")

    return [x - y for x, y in zip(a, b)]

# Calculate and display the results
print("Vector a:", a)
print("Vector b:", b)
print("Addition:", add_vectors(a, b))
print("Subtraction:", subtract_vectors(a, b))
```

## 4. Scalar multiplication

A **scalar** is a single number. Multiplying every component of a vector by scalar $c$ gives

$$
c\mathbf{a}=[ca_1,ca_2,\ldots,ca_n].
$$

For $\mathbf{a}=[2,3]$ and $c=3$:

$$
3\mathbf{a}=[6,9].
$$

A negative scalar reverses the vector's direction; a scalar with absolute value greater than 1 increases its magnitude. Multiplication by zero produces the zero vector.

### Your turn

Implement `scale_vector(scalar, vector)` with pure Python. Test a positive scalar, a negative scalar, and zero.

```python
def scale_vector(scalar, vector):
    """Multiply every component of a vector by a scalar."""
    return [scalar * component for component in vector]

# Test 1: Positive scalar
print("Positive scalar:", scale_vector(3, [2,3]))

# Test 2: Negative scalar
print("Negative scalar:", scale_vector(-2, [2,3]))

# Test 3: Zero scalar
print("Zero scalar:", scale_vector(0, [2,3]))
```

## 5. Dot product

For two vectors of the same dimension, their dot product is the sum of pairwise products:

$$
\mathbf{a}\cdot\mathbf{b}=\sum_{i=1}^{n}a_i b_i.
$$

For $\mathbf{a}=[2,3]$ and $\mathbf{b}=[1,4]$:

$$
\mathbf{a}\cdot\mathbf{b}=(2)(1)+(3)(4)=14.
$$

The result is a **scalar**, not a vector.

### Geometric interpretation

For non-zero vectors, the dot product also satisfies

$$
\mathbf{a}\cdot\mathbf{b}=\|\mathbf{a}\|_2\,\|\mathbf{b}\|_2\cos\theta,
$$

where $\theta$ is the angle between them. A positive dot product means the angle is acute, zero means the vectors are orthogonal, and a negative value means the angle is obtuse.

### Your turn

Implement `dot_product(a, b)` using pure Python. Include dimension validation and test a pair whose dot product is zero.

```python
def dot_product(a, b):
    """Calculate the dot product of two vectors."""

    if len(a) != len(b):
        raise ValueError("Vectors must have the same length.")

    return sum(x*y for x, y in zip(a, b))

# Test 1: Worked example
a = [2, 3]
b = [1, 4]

print("Dot product:", dot_product(a, b))

# Test 2: Orthogonal vectors
u = [1, 0]
v = [0, 1]

print("Orthogonal vectors dot product:", dot_product(u, v))
```

## 6. Euclidean magnitude (L2 norm)

The Euclidean norm measures vector length:

$$
\|\mathbf{v}\|_2=\sqrt{v_1^2+v_2^2+\cdots+v_n^2}.
$$

For $\mathbf{a}=[2,3]$:

$$
\|\mathbf{a}\|_2=\sqrt{2^2+3^2}=\sqrt{13}.
$$

A norm is always non-negative, and only the zero vector has zero Euclidean norm.

### Your turn

Implement `magnitude(vector)` using pure Python. Check that the magnitude of `[3, 4]` is 5. You should also decide and document how your function handles an empty sequence.

```python
from math import sqrt


def magnitude(vector):
    """Calculate the Euclidean magnitude of a vector."""

    if len(vector) == 0:
        raise ValueError("Cannot calculate magnitude of an empty vector.")

    return sqrt(sum(component ** 2 for component in vector))


# Test 1: Worked example
print("Magnitude of [3, 4]:", magnitude([3, 4]))

# Test 2: Another vector
print("Magnitude of [2, 3]:", magnitude([2, 3]))

# Test 3: Empty vector
try:
    magnitude([])
except ValueError as error:
    print("Empty vector:", error)
# Check your result against the [3, 4] example.
# Add a deliberate policy for an empty vector and document that choice.
```

## 8. Cosine similarity: a machine-learning connection

For two non-zero vectors, cosine similarity is the cosine of the angle between them:

$$
\operatorname{cosine\_similarity}(\mathbf{a},\mathbf{b})=
\frac{\mathbf{a}\cdot\mathbf{b}}{\|\mathbf{a}\|_2\,\|\mathbf{b}\|_2}.
$$

For real-valued, non-zero vectors, the result lies between -1 and 1. Values closer to 1 indicate similar directions; values near 0 indicate orthogonal directions; values near -1 indicate opposite directions.

Cosine similarity is used in applications such as comparing text embeddings in semantic search. It compares direction rather than raw magnitude.

**Edge case:** the expression is undefined if either input is a zero vector, because the denominator is zero. Your implementation should have a clear policy for this case, such as raising `ValueError`.

### Your turn

Implement cosine similarity using your own dot-product and magnitude functions. Verify it with NumPy. Add a test that confirms your chosen zero-vector policy.

```python
def cosine_similarity(a, b):
    """Calculate cosine similarity using our own vector functions."""

    if len(a) != len(b):
        raise ValueError("Vectors must have the same dimension.")

    magnitude_a = magnitude(a)
    magnitude_b = magnitude(b)

    if magnitude_a == 0 or magnitude_b == 0:
        raise ValueError("Cosine similarity is undefined for zero vectors.")

    return dot_product(a, b) / (magnitude_a * magnitude_b)


# Test with the worked example
a = [2, 3]
b = [1, 4]

python_similarity = cosine_similarity(a, b)

print("Cosine similarity:", python_similarity)
```

## 10. Review and Exit Ticket — Vectors

### 1. Difference between dimension and magnitude

The dimension of a vector tells us how many components it contains.
The magnitude measures its length, calculated using the Euclidean norm.

For example, the vector [3, 4] has dimension 2 and magnitude 5.

### 2. Python list addition vs NumPy array addition

Python list addition concatenates two lists, whereas NumPy array
addition adds corresponding elements.

For example:
- [2, 3] + [1, 4] in Python gives [2, 3, 1, 4].
- NumPy array addition gives [3, 7].

### 3. Why must dot-product inputs have compatible dimensions?

The dot product multiplies corresponding components and adds
the products. Therefore, two one-dimensional vectors must contain
the same number of components.

### 4. Difference between NumPy shapes (2,) and (2, 1)

The shape (2,) represents a one-dimensional array containing
two elements.

The shape (2, 1) represents a two-dimensional array with
two rows and one column.

### 5. Why is cosine similarity undefined for a zero vector?

Cosine similarity divides the dot product by the product
of the two vector magnitudes. A zero vector has magnitude zero,
which makes the denominator zero. Therefore, the formula
is undefined when either vector is a zero vector.

### 6. One ML/AI application of vectors

Vectors represent numerical features and learned embeddings.
In semantic search, cosine similarity can compare text embeddings
to find texts that have similar directions in the embedding space.
