## Exercise 1 — Component-wise operations

Given

$$
\mathbf{a}=[4,-2,7], \qquad \mathbf{b}=[1,5,3],
$$

1. Calculate $\mathbf{a}+\mathbf{b}$, $\mathbf{a}-\mathbf{b}$, and $2\mathbf{a}$ by hand.
2. Implement addition, subtraction, and scalar multiplication with Python lists.
3. Verify the results with NumPy.

```python
# Define the vectors
a = [4, -2, 7]
b = [1, 5, 3]

# Validate dimensions
if len(a) != len(b):
    raise ValueError("Vectors must have the same dimension.")

# Component-wise operation using pure Python
addition = [x + y for x, y in zip(a, b)]
subtraction = [x - y for x, y in zip(a, b)]
scalar_multiplication = [2 * x for x in a]

# Display results
print("Vector a:", a)
print("Vector b:", b)
print("Addition:", addition)
print("Subtraction:", subtraction)
print("2a:", scalar_multiplication)
```

```python
import numpy as np

a_np = np.array(a)
b_np = np.array(b)

np_addition = a_np + b_np
np_subtraction = a_np - b_np
np_scalar_multiplication = 2 * a_np

print("NumPy addition:", np_addition)
print("NumPy subtraction:", np_subtraction)
print("NumPy scalar multiplication:", np_scalar_multiplication)

print("\n--- Verification ---")
print("Addition matches:", np.allclose(addition, np_addition))
print("Subtraction matches:", np.allclose(subtraction, np_subtraction))
print(
    "Scalar multiplication matches:",
    np.allclose(scalar_multiplication, np_scalar_multiplication)
)
```

## Exercise 2 — Dot product and magnitude

Given

$$
\mathbf{u}=[1,2,3], \qquad \mathbf{v}=[4,-1,2],
$$

1. Calculate $\mathbf{u}\cdot\mathbf{v}$ manually.
2. Calculate $\|\mathbf{u}\|_2$.
3. Implement both operations with pure Python.
4. Compare your results with `np.dot` and `np.linalg.norm`.

```python

from math import sqrt


# Define the vectors
u = [1, 2, 3]
v = [4, -1, 2]


def dot_product(u, v):
    """Calculate the dot product of two vectors."""
    if len(u) != len(v):
        raise ValueError("Vectors must have the same dimension.")

    return sum(x * y for x, y in zip(u, v))


def magnitude(vector):
    """Calculate the Euclidean magnitude of a non-empty vector."""
    if len(vector) == 0:
        raise ValueError("Vector must contain at least one component.")

    return sqrt(sum(component ** 2 for component in vector))


# Calculate results using pure Python
python_dot = dot_product(u, v)
python_magnitude = magnitude(u)

print("Vector u:", u)
print("Vector v:", v)
print("Dot product:", python_dot)
print("Magnitude of u:", python_magnitude)
```

```python

import numpy as np

u_np = np.array(u, dtype=float)
v_np = np.array(v, dtype=float)

numpy_dot = np.dot(u_np, v_np)
numpy_magnitude = np.linalg.norm(u_np)

print("NumPy dot product:", numpy_dot)
print("NumPy magnitude:", numpy_magnitude)

print("\n--- Verification ---")
print("Dot product matches:", np.isclose(python_dot, numpy_dot))
print(
    "Magnitude matches:",
    np.isclose(python_magnitude, numpy_magnitude)
)
```

```python
# Test: vectors with incompatible dimensions
try:
    dot_product([1, 2, 3], [4, 5])
except ValueError as error:
    print("Dimension check:", error)


# Test: empty vector
try:
    magnitude([])
except ValueError as error:
    print("Empty-vector check:", error)
```

## Exercise 3 — Dimension mismatch and validation

Implement a validation rule that rejects vectors of different lengths.

Explain why standard component-wise addition of a 2-component vector and a 3-component vector is not defined. Your implementation should raise a clear `ValueError` when the dimensions do not match.

### Why dimension validation is necessary

Standard component-wise vector addition requires both vectors to
have the same number of components.

If one vector has two components and another has three, there is
no corresponding component for every position. Therefore, the
operation is not defined as standard vector addition.

A validation check prevents invalid inputs from being processed
silently and raises a clear ValueError instead.

```python

def add_vectors_checked(a, b):
    """Add vectors after validating their dimensions."""

    if len(a) != len(b):
        raise ValueError("Vectors must have the same dimension.")

    return [x + y for x, y in zip(a, b)]


# Test 1: Matching dimensions
a = [1, 2]
b = [3, 4]

result = add_vectors_checked(a, b)
print("Valid addition:", result)


# Test 2: Mismatched dimensions
a = [1, 2]
b = [3, 4, 5]

try:
    result = add_vectors_checked(a, b)
except ValueError as error:
    print("Invalid addition rejected:", error)
```

## Exercise 4 — Cosine similarity

For

$$
\mathbf{p}=[1,2,3], \qquad \mathbf{q}=[2,1,3],
$$

1. Implement cosine similarity from its mathematical definition.
2. Compare your output with a NumPy calculation.
3. Decide how your function should behave when one input is a zero vector, and implement that policy.

```python

def cosine_similarity(p, q):
    """Calculate cosine similarity using our own vector functions."""

    if len(p) != len(q):
        raise ValueError("Vectors must have the same dimension.")

    norm_p = magnitude(p)
    norm_q = magnitude(q)

    if norm_p == 0 or norm_q == 0:
        raise ValueError(
            "Cosine similarity is undefined for zero vectors."
        )

    return dot_product(p, q) / (norm_p * norm_q)


# Given vectors
p = [1, 2, 3]
q = [2, 1, 3]

python_cosine = cosine_similarity(p, q)

print("Vector p:", p)
print("Vector q:", q)
print("Cosine similarity:", python_cosine)
```

```python

import numpy as np

p_np = np.array(p, dtype=float)
q_np = np.array(q, dtype=float)

numpy_cosine = (
    np.dot(p_np, q_np)
    / (np.linalg.norm(p_np) * np.linalg.norm(q_np))
)

print("Python result:", python_cosine)
print("NumPy result:", numpy_cosine)
print("Results match:", np.isclose(python_cosine, numpy_cosine))
```

```python
try:
    cosine_similarity([0, 0, 0], [1, 2, 3])
except ValueError as error:
    print("Correctly handled:", error)
```

## Exercise 5 — Higher-dimensional vectors and reproducibility

Generate two vectors with at least 100 components using a seeded NumPy random generator.

- Calculate their dot product using pure Python.
- Verify the result with NumPy.
- Explain why a fixed seed helps make an experiment reproducible.
- Use `np.isclose` rather than assuming floating-point results are always exactly equal.

```python

import numpy as np

# Step 1: Create a seeded random generator
rng = np.random.default_rng(seed=42)

# Step 2: Generate two vectors with 100 components each
a_np = rng.normal(size=100)
b_np = rng.normal(size=100)

# Step 3: Convert NumPy arrays to Python lists
a = a_np.tolist()
b = b_np.tolist()

print("Number of components in a:", len(a))
print("Number of components in b:", len(b))

# Step 4: Calculate dot product using pure Python
python_dot = sum(x * y for x, y in zip(a, b))

# Step 5: Verify using NumPy
numpy_dot = np.dot(a_np, b_np)

print("\nPure Python dot product:", python_dot)
print("NumPy dot product:", numpy_dot)
print("Results match:", np.isclose(python_dot, numpy_dot))
```

```python

# Create another generator with the same seed
rng_repeat = np.random.default_rng(seed=42)

# Generate the same sequence again
a_repeat = rng_repeat.normal(size=100)
b_repeat = rng_repeat.normal(size=100)

print("Vector a reproduced:", np.array_equal(a_np, a_repeat))
print("Vector b reproduced:", np.array_equal(b_np, b_repeat))
```

### 1. What is the difference between vector dimension and magnitude?

A vector's dimension tells us how many components it contains,
whereas its magnitude measures its length.

For example, the vector [3, 4] has dimension 2 and Euclidean
magnitude 5.

### 2. What does NumPy shape (n,) indicate?

The shape (n,) indicates a one-dimensional NumPy array
containing n elements.

For example, an array with shape (3,) has three elements
along one axis. Its `ndim` is 1, while its `size` is 3.

### 3. Give one application of cosine similarity in AI.

Cosine similarity can compare text embeddings in semantic search.
A search system can compare the embedding of a user's query
with stored document embeddings to find texts with similar
directions in the embedding space.
